### Outlier detection

```python
# X--> Input features and Y--> Target values
x = df.drop(columns='Outcome',axis=1)
y = df['Outcome']
```

```python
fig,ax=plt.subplots(figsize=(15,15))
sns.boxplot(data=x,ax=ax)
plt.show()
```

```python
cols= ['Insulin', 'Pregnancies', 'Glucose', 'BloodPressure', 'SkinThickness', 'BMI','Age','DiabetesPedigreeFunction']
for col in cols:
    Q1=x[col].quantile(0.25)
    Q3=x[col].quantile(0.75)
    IQR=Q3-Q1
    lower_bound = Q1-1.5*IQR
    upper_bound= Q3+1.5*IQR
    mask = (x[col]>= lower_bound) & (x[col]<= upper_bound)
```

```python
x_outlier_detection = x[mask]
y_outlier_detection = y[mask]
```

```python
from sklearn.preprocessing import StandardScaler
scaler =StandardScaler()
x_scaler = scaler.fit_transform(x_outlier_detection)
```

```python
fig, ax = plt.subplots(figsize=(15,15))
sns.boxplot(data=x_scaler,ax=ax)
```

```python
fig, ax = plt.subplots(figsize=(15,15))
sns.boxplot(data=datanew,ax=ax)
```

#### Model training

#### Splitting of data into training and testing

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(datanew,y_outlier_detection,test_size=0.33,random_state=42)
```

```python
x_train.shape
```

```python
x_test.shape
```

#### Data imbalancing

```python
y_train.value_counts()
```

#### SMOTE (SYNTHETIC MINORITY OVERSAMPLING TECHNIQUE)

```python
from imblearn.over_sampling import SMOTE
smote = SMOTE (random_state=42)
x_train_resampled, y_train_resampled = smote.fit_resample(x_train,y_train)
# Check resampled class distribution
print("\nResamp1ed class distribution: ")
print (pd. Series (y_train_resampled).value_counts())
```

```python
from sklearn.linear_model import LogisticRegression
classification = LogisticRegression()
classification.fit(x_train_resampled,y_train_resampled)
```

#### Model Predictions

```python
y_predictions = classification.predict(x_test)
print(y_predictions)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_predictions)
```

```python
from sklearn.metrics import classification_report
target_names = ['Non-diabetic','Diabetic']
print(classification_report(y_test,y_predictions,target_names=target_names))
```
